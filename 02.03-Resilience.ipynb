### Resilience in the sustainability sciences

Resilience as a **metaphor related to sustainability**

Resilience and sustainability are closely related concepts in the context of social-ecological systems (SES). Resilience refers to the capacity of a system to absorb disturbances, adapt to changes, and maintain its core functions and structures. Sustainability, on the other hand, is the ability to meet the needs of the present without compromising the ability of future generations to meet their own needs. A resilient system can be more sustainable because it can withstand and adapt to shocks and stresses, ensuring long-term stability and functionality. Therefore, enhancing resilience is often seen as a key strategy for achieving sustainability.

Resilience as a **property of dynamic systems**

Dynamic systems are those that change over time, often in response to internal or external stimuli. Resilience in these systems is about how well they can absorb shocks and continue to operate effectively. For example, an ecosystem might experience a natural disaster but still maintain its biodiversity and functionality.



Resilience as a **measurable quantity**


Resilience is considered a measurable quantity through various indicators and metrics that capture the capacity of systems to absorb disturbances, adapt to changes, and maintain functionality. In field studies of social-ecological systems (SES), resilience can be assessed using indicators related to ecological, social, economic, and institutional dimensions. These indicators help researchers quantify resilience and understand how different systems respond to various stressors and shocks.

> It is important to **acknowledge these different meanings** of resilience when discussing sustainability and human-environment interactions.


## Conceptualizing resilience

### Resilience vs. dynamics

Resilience differs from merely being static or unchanging over time; resilient systems are often quite dynamic. Conversely, systems that remain constant over time can lack resilience. Acknowledging the **difference between static stability and resilience** is crucial [@Meadows2009].

* **Static stability** is observable; it can be assessed by analyzing changes in the system's conditions over time.
* **Resilience**, on the other hand, is often only noticeable when the system is pushed beyond its limits and breaks down. Because resilience may not be apparent without a systems view, individuals frequently prioritize stability, productivity, or other more immediately observable characteristics over resilience.


For example, **just-in-time deliveries** of products to retailers and parts to manufacturers have minimized inventory fluctuations and lowered costs across various industries. Nonetheless, this mode of operation has rendered the production system more vulnerable to disruptions in fuel supply, computer failures, labor shortages, and other potential shocks. 

Another example constitutes the **intensive management of European forests**. Over centuries, it has transformed native ecosystems into single-age, single-species plantations, frequently composed of nonnative trees. These plantations aim to produce wood and pulp consistently over time. However, these forests have lost their resilience without multiple species interacting with each other and their environment. As a result, we are witnessing their vulnerability to threats such as industrial air pollution and pests like the bark beetle.

### Resilience of what to what?

The resilience of what to what is **a key question** when applying the concept of resilience to sustainability and human-environment interactions [@CarpenterEtAl2001].

The **'of what'** refers to system function or configuration to be sustained, such as biodiversity, ecosystem services, social cohesion, or economic stability. 

The **'to what'** refers to the disturbances, shocks, or changes that the system needs to withstand, such as climate change, natural disasters, economic crises, or social conflicts.

### Specified vs. general resilience

Another key distinction in resilience research is between specified and general resilience [@FolkeEtAl2010].

**Specified resilience** refers to the resilience of a system function to specific challenges or disturbances (i.e., a narrowly defined *what* to *what*). *For example*, 
- a community might have specified resilience to flooding by building dams and flood protection walls;
- a farmer might use pest-resistant crops to increase resilience to specific pest infestations; or
- an individual might get vaccinated against a particular disease to increase resilience to that disease.

**General resilience** refers to the system's capacity to deal with the unknown, uncertainty, and surprise (i.e., a broadly defined *what* to *what*). *For example*,
- a community or society with functioning institutions and social networks has generalized resilience to various shocks and stresses;
- a farmer with a healthy business model and diversified crops has generalized resilience to various economic and environmental changes;
- a functioning immune system can provide generalized resilience to a wide range of diseases;

### Resilience types or capacities

While resilience, in general, is defined as the capacity of a system to absorb disturbances, adapt to changes, and maintain functionality, it is useful to differentiate between three types of resilience that can be distinguished based on the system's response to stressors and shocks:

- **Robustness** is the capacity to resist (or absorb) change and continue to function in the present state.
- **Adaptation** is the capacity to adjust responses to changing external drivers and continue developing within the current basin of attraction [@FolkeEtAl2010].
- **Transformation** is the capacity to cross thresholds into new development trajectories [@FolkeEtAl2010][^transformation].

[^transformation]: This definition matches the ball-and-cup diagram of @MeuwissenEtAl2019. @WalkerEtAl2004 define transformability more broadly, as the capacity to create a fundamentally new system when ecological, economic, or social structures make the existing system untenable. For them, a new system has a new stability landscape with new state variables. The model in this lecture has a fixed state variable and cannot represent such a change.

These three types are often illustrated by ball-and-cup diagrams

A lot of resilience scholarship utilizes **qualitative methods, case studies, and conceptual frameworks** (i.e., mental, verbal and pictorial models) to understand the dynamics of social-ecological systems. While these approaches are invaluable for generating diverse insights and hypotheses, they have **difficulty** in providing a **precise understanding** of resilience that allows for **quantitative predictions** and **generalizable results** in the sense of identifying universal system structures of relevance.


Here, the mathematics of stochastic dynamics and bifurcations can help.

## Operationalizing resilience

What happens to the tipping element of the previous lecture when it is exposed to **random shocks**, one-off random changes of its state? We add shocks to the model from [Tipping Elements](https://wbarfuss.github.io/csm-of-hei/02.02-TippingElements.html),

$$x_{t+1} = x_t + \frac{1}{\tau}\left(h + rx_t - x_t^3\right) + n\eta_t,$$

where $h$ is the external influence, $r$ the strength of the reinforcing feedback loop relative to the balancing feedback loop, and $\tau$ the response time of the system. The shocks $\eta_t$ are normally distributed with mean zero and unit variance; $n$ sets their strength. The shocks come from outside the system, so they enter outside the bracket.

In Python, the update without and with shocks is,

```
def F_tipmod(x, h, r, tau): return x + (h + r*x - x**3)/tau
def F_tipmod_noise(x, h, r, tau, n, rng): 
    return F_tipmod(x, h, r, tau) + n*rng.standard_normal(np.shape(x))
```

The equilibrium points are the real roots of $h + rx - x^3 = 0$. For $|h| < h_{crit}$, there are three, sorted from the lower cup over the edge to the upper cup.

```
def equilibria(h, r=1):  # real roots of h + r*x - x**3 = 0, sorted
    roots = np.roots([-1, 0, r, h])  # polynomial coefficients
    return np.sort(roots[np.abs(roots.imag) < 1e-9].real) # keep only real part

equilibria(h=-0.2)
```

From [Tipping Elements](https://wbarfuss.github.io/csm-of-hei/02.02-TippingElements.html), we take the potential function and the bifurcation diagram. The range of $h$ in the bifurcation diagram and the range of the potential grow with $r$.

```
def G_tipmod(x, h,r,tau): return - (h*x + r*x**2/2 - x**4/4)/tau

def plot_tipmod_potential(h=0.2, r=1.0, tau=2):
    x_grid=np.linspace(-2,2,501); plt.ylim(min(-0.5, G_tipmod(x_grid, h,r,tau).min()-0.05), 0.5)
    plt.plot(x_grid, G_tipmod(x_grid, h,r,tau), color='blue')
    plt.ylabel(r'Potential $G(x)$'); plt.xlabel(r'System state $x$')

    xeq = equilibria(h, r)  # equilibrium points
    plt.plot(xeq, G_tipmod(xeq, h, r, tau), 'o', ms=12, color='k')

def plot_equilibrium_points_tipmod(r=1.0):
    xe=np.linspace(-2.0,2.0,501) # equilibrium points
    h = xe**3 - r*xe # parameter h
    plt.plot(h, xe, "--"); # plot
    plt.xlabel(r'External influence $h$'); plt.ylabel(r'Equilibrium points $x_e$');

def plot_stability_tipmod(r=1.0, tau=3.0, hextent=[-3.0,3.0]):
    xe=np.linspace(-3, 3, 1001) # equilibrium points
    h = xe**3 - r*xe # parameter h

    def F_(x, r,tau): return 1 + (r-3*x**2)/tau
    cond=np.logical_and(F_(xe, r,tau)<1, F_(xe, r,tau)>-1)
    plt.plot(h[cond], xe[cond], ".", c='red')

    plt.xlabel(r'External influence $h$'); plt.ylabel(r'Equilibrium points $x_e$');
    plt.xlim(hextent);

def plot_analytical_bifurcation_tipmod(r = 1.0, tau = 3.0, extent=3.0):
    plot_equilibrium_points_tipmod(r=r);
    plot_stability_tipmod(r=r, tau=tau, hextent=[-extent, extent]);

def h_crit(r=1): return 2*r/3*np.sqrt(r/3)  # critical value of h (see Tipping Elements)
```

We simulate the system under shocks,

```
def plot_trajectory(h=-0.3, r=1, tau=2, n=0.0, xinit=0.5, iters=2500, seed=2):
    rng = np.random.default_rng(seed)  # random number generator; the seed makes the result reproducible
    x = xinit  # re-storing the initial values
    trajectory = [x]  # container to store the trajectory
    for t in range(iters):  # looping through the iterations
        x = F_tipmod_noise(x, h, r, tau, n, rng)
        if np.abs(x)>3: break  # stop the simulation when x becomes too large
        trajectory.append(x)  # storing the new state in the container
    plt.plot(trajectory, 'purple'); plt.ylim(-1.6, 1.6)
    plt.xlabel('Time steps $t$'); plt.ylabel('System state $x$');  # makes plot nice
    return np.array(trajectory)
```

and show the trajectory together with the potential and the bifurcation diagram.

```
def plot_all(h=-0.3, r=1, tau=2, n=0.0, xinit=0.5, seed=2):
    fig = plt.figure(figsize=(10,5))

    ax1 = fig.add_subplot(2,2,1)
    plot_tipmod_potential(h=h, r=r, tau=tau)

    ax2 = fig.add_subplot(2,2,2)
    plot_analytical_bifurcation_tipmod(r=r, tau=tau, extent=max(0.5, 1.3*h_crit(r)))
    ax2.plot([h, h], [-2, 2], 'k-') # include the value of h

    ax3 = fig.add_subplot(2,1,2)
    traj = plot_trajectory(h=h, r=r, tau=tau, n=n, xinit=xinit, seed=seed)

    # include trajectory in the potential and the bifurcation diagram, time from light to dark
    ax1.scatter(traj, G_tipmod(traj, h, r, tau), alpha=0.5, s=40, c=np.arange(len(traj)), cmap='plasma_r',zorder=10)
    ax2.scatter(np.ones_like(traj)*h, traj, alpha=0.5, s=40, c=np.arange(len(traj)), cmap='plasma_r',zorder=10)
    plt.tight_layout()
```

```
plot_all(h=-0.3, n=0.075, seed=2)
```

**The system leaves a cup that still exists.** A shock carries it over the edge into the other cup. In the deterministic model of the previous lecture, this was impossible.

### To what? | Pulse and press disturbances

**Resilience of what to what?** Two kinds of disturbances act on our model [@BenderEtAl1984]:

- A **pulse** disturbance is a one-off change of the system state $x$. In our model, each random shock $n\eta_t$ is a pulse, one per time step.
- A **press** disturbance is a sustained change of a parameter. In our model: a change of the external influence $h$.

A pulse can push the system over the edge of its cup (*noise-induced tipping*). A press can remove the cup altogether (*bifurcation-induced tipping*). Often, both act together (see the DeepDives).

**Of what?** At two levels. The **stability landscape**: which cup the system is in, and what the cups look like. The **current basin**: how the system stays in its cup.

### Of what? The potential | Robustness, adaptation, and transformation

The three resilience types concern the stability landscape as a whole. In our model:

- **Robustness**: the system stays in its cup despite the shocks.
- **Adaptation**: the cup changes. E.g., a larger $r$ makes both cups wider and deeper.
- **Transformation**: the system moves over the edge into another cup that maintains its function.

The resilience types concern the stability landscape as a whole: all cups and the position of the system among them. The measures in the next section concern only the cup the system is in.

Under robustness, the landscape stays as it is, and the system stays in its cup: disturbances move it within the cup, and it returns. Under adaptation, the cup changes around the current equilibrium point so that it can absorb larger disturbances. In our model, a stronger reinforcing loop $r$ makes the cup wider and deeper. Adaptation changes how strong the feedback loops are, not which feedback loops there are. The system stays in its cup. Under transformation, the system moves to another equilibrium state that maintains its function. A press can take it there by removing the current cup (bifurcation-induced tipping), a shock by pushing it over the edge (noise-induced tipping), or actors by moving it on purpose.

Whether a move to another cup is a transformation or a collapse does not follow from the dynamics. It depends on whether the new state maintains the function of the system, e.g., the income of a farm or the services of an ecosystem. Our model shows the state $x$, not the function. If only the upper cup provides the function, leaving it is a collapse. If the lower cup provides it as well, e.g., through a different way of farming, leaving the upper cup is a transformation. The same simulation can show either.

Adaptation makes the current cup wider and deeper, whether the current state is desirable or not. A transformation, in contrast, requires the system to leave its cup, which is easier the less resilient the cup is. Robustness and adaptation thus work against transformation. More resilience is not always better.

Increase the shock strength $n$, then adapt the system by increasing $r$. Keep in mind that $\tau$ has to grow with $r$: for $\tau < 2r$, the state overshoots its equilibrium point, and for $\tau < r$, the equilibrium point becomes unstable.

```
interact(plot_all, h=(-0.5, 0.5, 0.01), r=(0.5, 2, 0.1), tau=(1, 10, 0.1),
         n=(0, 0.25, 0.005), xinit=fixed(0.5), seed=(0, 100, 1));
```

A larger $r$ stabilizes **whichever** state the system is in, including an undesirable one. Once the system has moved to the lower cup, adaptation locks it in. **Is more resilience always better?**

### Of what? The current cup

**Multiple measures exist** to quantify how well the system absorbs disturbances within its current basin. We will **focus on three**. They differ in the disturbance, i.e., in the resilience *to what*. Engineering resilience is the only linear measure; it holds only close to the equilibrium point. The other two concern the whole basin.

#### To what? A single pulse | Ecological resilience

**How does the system respond to a single pulse?** We consider the upper cup for $h=-0.2$: its bottom is the stable equilibrium point $x_s \approx 0.88$, its edge the unstable equilibrium point $x_u \approx 0.21$. We push the system with a pulse of size $s$ at time step $10$.

```
def plot_pulse(s, h=-0.2, r=1, tau=2, iters=40):
    x = equilibria(h, r)[-1]  # start at the bottom of the upper cup
    trajectory = [x]
    for t in range(iters):
        x = F_tipmod(x, h, r, tau)
        if t == 10: x = x - s  # the pulse
        trajectory.append(x)
    plt.plot(trajectory, label=f'pulse $s={s}$')
    plt.axhline(equilibria(h, r)[1], color='k', ls='--', lw=1)  # edge of the cup
    plt.xlabel('Time steps $t$'); plt.ylabel('System state $x$'); plt.legend()
```

```
plot_pulse(0.6); plot_pulse(0.75)
```

The system absorbs a pulse that keeps it in the cup. Otherwise, it tips into the lower cup. The **width** $x_s - x_u \approx 0.67$ of the cup is the largest downward pulse the system can absorb[^overshoot]. This is the **ecological resilience** of the system: the size of the disturbance it can absorb before it moves to another basin of attraction [@Holling1973].

[^overshoot]: In our discrete-time model, large upward pulses also tip the system. Far up the right wall of the cup, one time step throws the state past the bottom and over the edge. For $h=-0.2$ and $\tau=2$, this happens for upward pulses larger than $0.74$. The larger $\tau$, the larger this limit (see the DeepDive on $\tau$ in [Tipping Elements](https://wbarfuss.github.io/csm-of-hei/02.02-TippingElements.html)).

Real systems have such alternative basins. For example, in regions with the same intermediate rainfall, tree cover across tropical forests and savannas worldwide clusters around three values, suggesting three alternative stable states: forest, savanna, and treeless [@HirotaEtAl2011].



#### To what? A set of single random shocks | Basin stability

**Which share of random shocks does the system recover from?** This share is the **basin stability** [@MenckEtAl2013]. We estimate it by simulation: we hit the system at the upper equilibrium point with one normally distributed shock of strength $n$ and let it return without further shocks.

```
def basin_stability(h, n, r=1, tau=2, samples=10000, seed=0):
    rng = np.random.default_rng(seed)  # random number generator; the seed makes the result reproducible
    xs = equilibria(h, r)[-1]  # upper equilibrium point
    x = xs + n*rng.standard_normal(samples)  # one random shock per sample
    for _ in range(1000):  # deterministic return, without further shocks
        x = np.clip(F_tipmod(x, h, r, tau), -10, 10)  # clip: prevents overflow after very large shocks
    return np.mean(np.abs(x - xs) < 1e-3)  # share of samples back at the upper equilibrium point (diverged samples are not)
```

```
hs = np.linspace(-0.38, 0, 39)
for n in [0.3, 0.5]:
    plt.plot(hs, [basin_stability(h, n) for h in hs], 'o-', label=f'shock strength $n={n}$')
plt.xlabel('External influence $h$'); plt.ylabel('Basin stability'); plt.ylim(0, 1.05); plt.legend();
```

Basin stability depends on the **shocks**, i.e., on the resilience *to what*. It does not vanish at the tipping point: most shocks that push the system upward are still absorbed. In one dimension, basin stability follows from the width, apart from large upward shocks, which overshoot in our discrete-time model. Its strength lies in high-dimensional systems, where the shape of a basin cannot be drawn.

#### To what? A small pulse | Engineering resilience

Basin stability counts how often the system returns. **Engineering resilience** measures how fast it returns to the bottom of the cup after a pulse [@Pimm1984].

**How fast does the distance to the bottom shrink?** We write the state as $x_t = x_s + \hat x_t$ with a small distance $\hat x_t$. Close to $x_s$, the update function $F$ is approximately linear,

$$x_{t+1} = F(x_s + \hat x_t) \approx F(x_s) + F'(x_s)\,\hat x_t = x_s + F'(x_s)\,\hat x_t,$$

since $x_s$ is an equilibrium point, $F(x_s) = x_s$. Subtracting $x_s$ on both sides gives $\hat x_{t+1} \approx F'(x_s)\,\hat x_t$. After $t$ time steps,

$$|\hat x_t| \approx |F'(x_s)|^t\,|\hat x_0|, \qquad F'(x) = 1 + \frac{1}{\tau}(r - 3x^2).$$

Per time step, the system recovers the fraction $\frac{|\hat x_t| - |\hat x_{t+1}|}{|\hat x_t|} = 1 - |F'(x_s)|$ of its distance. This is the **return rate**.

We test the prediction: we push the system up by $\hat x_0 = 0.01$ and plot the distance to the bottom on a logarithmic axis, together with $|F'(x_s)|^t\,\hat x_0$.

```
def plot_return(h, hatx0=0.01, r=1, tau=2, iters=25):
    xs = equilibria(h, r)[-1]  # bottom of the upper cup
    x = xs + hatx0  # a small pulse upward
    distances = [hatx0]
    for t in range(iters):
        x = F_tipmod(x, h, r, tau); distances.append(np.abs(x - xs))
    Fp = np.abs(1 + (r - 3*xs**2)/tau)  # |F'(x_s)|
    line, = plt.plot(distances, 'o', label=f'Simulation, $h={h}$')
    plt.plot(hatx0*Fp**np.arange(iters+1), '-', color=line.get_color(),
             label=f"$|F'(x_s)|^t \hat x_0$, $|F'(x_s)|={Fp:.2f}$")
    plt.yscale('log'); plt.xlabel('Time steps $t$');
    plt.ylabel('Distance $|x_t - x_s|$'); plt.legend()

plot_return(-0.2); plot_return(-0.35)
```

The simulation follows the prediction: on a logarithmic axis, the distance falls along a straight line, by the factor $|F'(x_s)|$ per time step. Closer to the tipping point, $|F'(x_s)|$ is closer to $1$, and the system returns more slowly.

How do ecological and engineering resilience change under a press, when $h$ approaches the tipping point $-h_{crit}$?

```
def robustness_measures(h, r=1, tau=2):
    xe = equilibria(h, r)
    if len(xe) < 3: return np.nan, np.nan  # the upper cup has disappeared
    xu, xs = xe[1], xe[2]
    return xs - xu, 1 - np.abs(1 + (r - 3*xs**2)/tau)

hs = np.linspace(-0.45, 0, 451)
widths, rates = np.array([robustness_measures(h) for h in hs]).T
plt.plot(hs, widths, label='Width $x_s - x_u$'); plt.plot(hs, rates, label="Return rate $1-|F'(x_s)|$")
plt.axvline(-h_crit(), color='k', ls=':', lw=1)  # tipping point
plt.xlabel('External influence $h$'); plt.legend(); plt.ylim(0,1)
```

**Both vanish at the tipping point**, but they are different properties: the response time $\tau$ changes the return rate, not the width. Basin stability decreases as well. Under a press, the system becomes more vulnerable to shocks and can tip before the bifurcation is reached.

| Concept | Question | In our model |
|---|---|---|
| Pulse / press disturbance | Resilience to what? | random shocks $n\eta_t$ / change of $h$ |
| Robustness / adaptation / transformation | Stay in the cup / change the cup / move to another cup | $x$ near $x_s$ / change of $r$ / switch of cups |
| Ecological resilience | How large a single pulse can the system absorb? | width $x_s - x_u$ |
| Basin stability | Which share of random shocks does it absorb? | simulated share of returns |
| Engineering resilience | How fast does it return after a pulse? | return rate $1-\lvert F'(x_s)\rvert$ |

In a single system, measuring the width takes disturbances large enough to risk pushing it over the edge. The return rate can be observed from small shocks. **How can we detect the loss of resilience before the system tips?**

## Quantifying resilience

So far, we **operationalized multiple facets of resilience** through simple dynamical system models. While these models are crucial for understanding the underlying mechanisms of resilience, it is difficult to apply these models to real-world systems directly. 

For instance, critical questions around resilience, tipping elements, and regime shifts of real-world systems are **how resilient is the system?** and **how far away is the system from a tipping point?**. These questions are challenging to answer in empirical systems because we cannot and do not want to trigger the regime shift to find out. Tipping elements are often **hidden** and **uncertain**.

In the following, we will discuss how we can **quantify resilience** and thus **measure whether a system is approaching a tipping point**, based on the conceptual models we discussed above.

### Critical slowing down

Critical slowing down describes the phenomenon that the **internal time scale a system** operates on **increases**, when the system **approaches a tipping point**.  As a consequence, a system close to a tipping point tends to undergo **larger changes in response to perturbations** and takes longer to recover from them.

**How long does the return take?** We count the time steps $T$ until the distance has shrunk from $\hat x_0$ to a small threshold $\varepsilon$. With the result from the section on engineering resilience,

\begin{align}
|F'(x_s)|^T\,\hat x_0 &= \varepsilon \\
T \ln|F'(x_s)| &= \ln(\varepsilon/\hat x_0) \\
T &= \frac{\ln(\hat x_0/\varepsilon)}{-\ln|F'(x_s)|}.
\end{align}

Close to the tipping point, $|F'(x_s)|$ is close to $1$, and $-\ln|F'(x_s)| = -\ln\big(1 - (1-|F'(x_s)|)\big) \approx 1-|F'(x_s)|$, using the Taylor expansion $-\ln(1-y) \approx y + \mathcal O(y^2)$. Thus, the **return time is inversely proportional to the return rate**,

$$T \approx \frac{\ln(\hat x_0/\varepsilon)}{1-|F'(x_s)|}.$$

We simulate the return time for values of $h$ approaching the tipping point and compare it with the prediction.

```
def return_time(h, hatx0=0.01, eps=1e-6, r=1, tau=2, maxiter=100000):
    xs = equilibria(h, r)[-1]  # bottom of the upper cup
    x = xs + hatx0  # a small pulse upward
    for t in range(maxiter):
        if np.abs(x - xs) < eps: return t  # the system has returned
        x = F_tipmod(x, h, r, tau)
    return maxiter
```

```
hs = -h_crit() + np.geomspace(1e-4, 0.25, 40)  # values of h, denser close to the tipping point
return_rates = np.array([1 - np.abs(1 + (1 - 3*equilibria(h)[-1]**2)/2) for h in hs])  # 1-|F'(x_s)|

plt.plot(hs, [return_time(h) for h in hs], 'o', label='Simulation')
plt.plot(hs, np.log(0.01/1e-6)/return_rates, 'k-', label=r"$\ln(\hat x_0/\varepsilon)\,/\,(1-|F'(x_s)|)$")
plt.axvline(-h_crit(), color='k', ls=':', lw=1)  # tipping point
plt.yscale('log'); plt.xlabel('External influence $h$'); plt.ylabel('Return time $T$'); plt.legend();
```

The return time grows without bound as $h$ approaches $-h_{crit}$, because the return rate goes to zero. Close to the tipping point, the simulation follows the prediction. Further away, the prediction overestimates the return time, because $-\ln|F'(x_s)| \approx 1-|F'(x_s)|$ no longer holds. **Critical slowing down is the loss of engineering resilience**, seen as a return time instead of a return rate.

### Early-warning signals

Next, we use the phenomenon of critical slowing down as a **sign of resilience loss** to create an early-warning signal. Early-warning signals are important because they allow us to **anticipate critical transitions before they occur**. 

Early-warning signals are based on **statistical indicators** of the system behavior. Specifically, we will use the **autocorrelation** of the system's time series. Autocorrelation is the correlation of a signal with a delayed copy of itself as a function of the delay. It measures the degree of similarity between a given time series and a lagged version of itself. 

#### Conceptual model

We reuse our model with $r=1$, $\tau=2$, and shock strength $n=0.125$.

We create a synthetic time series, along which we slightly **reduce the resilience** of the upper equilibrium point by decreasing the external influence $h$ **with each time step**. Starting from $h=0$, we decrease $h$ by $0.00038$ per time step, so that it approaches the tipping point $-h_{crit} \approx -0.385$ after 1000 time steps.

```
def simulate_ramp(seed, xinit=1.1, iters=1000, n=0.125):
    rng = np.random.default_rng(seed)  # random number generator; the seed makes the result reproducible
    h = 0  # initial value of the external influence
    trajectory = []  # container to store the trajectory
    x = xinit  # re-storing the initial values
    for t in range(iters):  # looping through the iterations
        x_ = F_tipmod_noise(x, h, r=1, tau=2, n=n, rng=rng)
        trajectory.append(x_)  # storing the new state in the container
        x = x_  # the new state of the system `x_` becomes the current state `x`
        h -= 0.00038  # we slightly decrease h (i.e., reduce the resilience of the upper equilibrium point)
    return np.array(trajectory)

trajectory = simulate_ramp(seed=35)
```

```
plt.plot(trajectory, 'k'); plt.xlabel('Time step $t$'); plt.ylabel('System state $x$');
```

Now imagine, that we do not know the underlying model, but only have the time series. How can we detect the loss of resilience?

#### Scatter plot

To **visualize the autocorrelation**, (lag-1 temporal autocorrelation or AR(1) to be specific), we create a scatter plot of 200 points of our time series versus the 200 points of our time series at the next time step (lag-1). In this case, 200 is the size of our data **window**.

```
def scatter_autocorrelation(start=0):
    fig, axs = plt.subplots(1,2, figsize=(12,2.8))  # creates the two axes
    axs[0].scatter(trajectory[start:start+200], trajectory[start+1:start+201], s=10, c='blue');
    axs[0].set_xlabel(r'$x_{t}$'); axs[0].set_ylabel(r'$x_{t+1}$'); # makes first axis nice

    axs[1].plot(trajectory, 'k');  # plot the time series
    axs[1].fill_betweenx([-1.5, 1.25], [start, start], [start+200, start+200], color='blue', alpha=0.5) # show window
    axs[1].set_xlabel('Time steps $t$'); axs[1].set_ylabel(r'System state $x$'); axs[1].set_xlim(0, 1000) # makes axis nice
```

How do we see the **autocorrelation** in this plot? 

```
scatter_autocorrelation(start=0)
```

The autocorrelation is the **correlation** between the time series and a **lagged version** of itself. Thus, it is the correlation between the x-axis and the y-axis of the scatter plot on the left.

Then we **slide** our window of 200 system points **through our time series**. How does the scatter plot change? And what does that mean for the autocorrelation?

#### Autocorrelation

To quantify our visual understanding, we finally calculate the lag-1 temporal autocorrelation. 

For that, we use the `numpy.corrcoef` function. It returns the **correlation matrix** between the time series points from index 0 to 200 and from index 1 to 201. The autocorrelation is one of its off-diagonal elements,

```
np.corrcoef(trajectory[0:200], trajectory[1:201])[0,1]
```

**Sliding the window through the time series** until it ends at time step 800, before the transition at time step 873,

```
AR1 = [np.corrcoef(trajectory[start:start+200], 
                   trajectory[start+1:start+201])[0,1]
       for start in range(0, 600)]
```

```
plt.plot(np.arange(200, 800), AR1); plt.ylabel('AR(1)'); plt.xlabel('Time step $t$ (end of window)');
```

shows a clear rise of the lag-1 autocorrelation when approaching the tipping point, **indicating a loss of resilience**.

This method **requires only time series data**. It does not require knowledge about the exact system equations.

#### DeepDive | Why does the autocorrelation rise?

Close to the bottom of the cup, the distance $\hat x_t = x_t - x_s$ follows the linear update from the section on engineering resilience, now with shocks,

$$\hat x_{t+1} = F'(x_s)\,\hat x_t + n\eta_t.$$

The lag-1 autocorrelation is the covariance of successive distances divided by their variance (in a stationary state, both have the same variance). We multiply the update by $\hat x_t$ and average over time. The shock $\eta_t$ is independent of $\hat x_t$, so its term averages to zero,

\begin{align}
\mathrm{Cov}(\hat x_{t+1}, \hat x_t) &= \mathrm{Cov}(F'(x_s) \hat x_{t} + n\eta_t, \hat x_t) \\
&= F'(x_s)\,\mathrm{Var}(\hat x_t) + \mathrm{Cov}(n\eta_t, \hat x_t) \\
&= F'(x_s)\,\mathrm{Var}(\hat x_t)
\end{align}

Dividing by $\mathrm{Var}(\hat x_t)$ gives $\mathrm{AR}(1) = F'(x_s)$.

Along our ramp, $F'(x_s) \geq 0$. Thus,

$$\mathrm{AR}(1) = F'(x_s) = 1 - \big(1 - |F'(x_s)|\big),$$

one minus the return rate. As the return rate goes to zero at the tipping point, AR(1) goes to one. **The autocorrelation measures the return rate from fluctuations alone.**

#### DeepDive | Variance

The same update predicts how large the fluctuations are. We take the variance of both sides. The two terms are independent, and $\eta_t$ has unit variance,

$$\mathrm{Var}(\hat x_{t+1}) = F'(x_s)^2\,\mathrm{Var}(\hat x_t) + n^2.$$

In a stationary state, $\mathrm{Var}(\hat x_{t+1}) = \mathrm{Var}(\hat x_t)$. Solving for it, step by step,

\begin{align}
\mathrm{Var}(\hat x_t)\,\big(1 - F'(x_s)^2\big) &= n^2 \\
\mathrm{Var}(\hat x_t) &= \frac{n^2}{1 - F'(x_s)^2}.
\end{align}

As $F'(x_s)$ approaches $1$, the variance grows without bound. The **variance** is a second early-warning signal, and it also follows from the return rate.

We compare both indicators with their predictions, evaluating $F'(x_s)$ at the value of $h$ in the middle of each window.

```
def F_prime_s(h, r=1, tau=2): return 1 + (r - 3*equilibria(h, r)[-1]**2)/tau  # F'(x_s) at the bottom of the upper cup

ends = np.arange(200, 800)  # time step at the end of each window
Fp = np.array([F_prime_s(-0.00038*(end - 100)) for end in ends])  # h in the middle of each window
variances = [np.var(trajectory[end-200:end+1]) for end in ends]  # same windows as for AR(1)
```

```
fig, axs = plt.subplots(1, 2, figsize=(10, 2.8))
axs[0].plot(ends, AR1, label='Measured'); axs[0].plot(ends, Fp, 'k--', label="$F'(x_s)$")
axs[0].set_xlabel('Time step $t$ (end of window)'); axs[0].set_ylabel('AR(1)'); axs[0].legend()
axs[1].plot(ends, variances, label='Measured'); axs[1].plot(ends, 0.125**2/(1 - Fp**2), 'k--', label="$n^2/(1-F'(x_s)^2)$")
axs[1].set_xlabel('Time step $t$ (end of window)'); axs[1].set_ylabel('Variance'); axs[1].legend()
plt.tight_layout()
```

Both indicators follow their predictions, and both rise because the return rate falls. AR(1) rises clearly, from about $0$ to about $0.55$. The predicted variance rises by only about a third over the same time, less than its fluctuations from window to window. The variance grows strongly only when $F'(x_s)$ is close to $1$, and it also depends on the shock strength $n$, which is often unknown. AR(1) measures $F'(x_s)$ directly.

In this realization, the measured AR(1) lies below the prediction until time step 400 and above it afterwards. Averaged over many realizations, it lies about $0.08$ above the prediction. Two effects raise it: the bottom $x_s$ of the cup moves within each window, and the strong shocks carry the system beyond the range of the linear approximation.

#### DeepDive | Robustness of early-warning signals

Critical slowing down detects the slow loss of resilience under a press, before a bifurcation. It does not warn of a single large pulse or of a sudden fast change (see the DeepDive on three ways to tip below).

The indicator is also statistical. In the lecture, we showed one realization of the shocks (seed 35). **How typical is it?** We simulate 200 realizations of the same ramp of $h$. For each, we record the time step at which the system tips (its state first drops below $0$) and compute AR(1) in all windows that end at least 10 time steps before the tipping.

```
def ar1_before_tipping(trajectory, window=200, margin=10):
    tip = np.argmax(trajectory < 0) if np.any(trajectory < 0) else len(trajectory)  # first time step in the lower cup
    ends = np.arange(window, tip - margin)  # windows that end at least `margin` time steps before the tipping
    AR1 = np.array([np.corrcoef(trajectory[e-window:e], trajectory[e-window+1:e+1])[0,1] for e in ends])
    return tip, ends, AR1

runs = [ar1_before_tipping(simulate_ramp(seed)) for seed in range(200)]
tips = np.array([tip for tip, ends, AR1 in runs])  # time steps of tipping
rises = np.array([AR1[-1] - AR1[0] for tip, ends, AR1 in runs])  # rise from the first to the last window
```

```
fig, axs = plt.subplots(1, 2, figsize=(10, 2.8))
for run_tip, run_ends, run_AR1 in runs:
    axs[0].plot(run_ends, run_AR1, color='purple', alpha=0.1, lw=0.5)  # one line per realization
axs[0].plot(runs[35][1], runs[35][2], color='k', lw=1, label='seed 35 (lecture)')
axs[0].axvline(h_crit()/0.00038, color='k', ls=':', lw=1)  # time step at which h reaches -h_crit
axs[0].set_xlabel('Time step $t$ (end of window)'); axs[0].set_ylabel('AR(1)'); axs[0].legend(loc='upper left')
axs[1].scatter(tips, rises, s=10, c='purple', alpha=0.6)
axs[1].scatter(tips[35], rises[35], s=40, c='k'); axs[1].annotate(' seed 35 (lecture)', (tips[35], rises[35]))
axs[1].axhline(0, color='k', lw=0.5)
axs[1].set_xlabel('Time step of tipping'); axs[1].set_ylabel('Rise of AR(1)')
plt.tight_layout()
```

```
print(f'Median time step of tipping: {np.median(tips):.0f}, at h = {-0.00038*np.median(tips):.2f}')
print(f'Share of realizations with a rise of AR(1) above 0.2: {np.mean(rises > 0.2):.2f}')
print(f'Share of realizations with a larger rise than seed 35: {np.mean(rises > rises[35]):.2f}')
```

**All realizations tip before the bifurcation**, which $h$ would reach at time step $h_{crit}/0.00038 \approx 1013$ (dotted line); each line on the left ends 10 time steps before its realization tips. Half of them tip before time step 756, at $h \approx -0.29$. The shocks trigger the transition while the upper cup still exists (noise-induced tipping, see the DeepDive on three ways to tip).

**AR(1) rises in almost all realizations**, in about 9 of 10 by more than $0.2$ (right). The spread between realizations is large (left): at a given time step, the middle 80% of realizations span about $0.25$ in AR(1), while the median rises from $0.2$ at time step 300 to $0.5$ at time step 700. A single value of AR(1) therefore says little; its trend over time carries the warning. The realization in the lecture, seed 35, is a favorable example: only 8% of the realizations rise more. The rise is smaller in realizations that tip early: they tip further from the bifurcation, where critical slowing down is weaker. The early-warning signal is thus robust to the realization of the shocks, but its strength depends on how close to the bifurcation the system tips.

Last, a trend in the time series also raises the autocorrelation. In applications, we therefore remove the trend from each window before computing AR(1). We skipped this step here.

## DeepDive | Three ways to tip

@AshwinEtAl2012 distinguish three ways in which a system can leave its basin of attraction:

- **Bifurcation-induced tipping**: a parameter crosses a bifurcation, and the cup disappears.
- **Noise-induced tipping**: a shock carries the system over the edge of a cup that still exists.
- **Rate-induced tipping**: external conditions change faster than the system can follow, although a cup exists at every moment.

All three occur in our model. Real transitions often combine them: a press brings the system close to the bifurcation, and a shock completes the transition.

### Bifurcation-induced tipping

We decrease $h$ slowly from $0$ to $-0.5$ without shocks. The system follows the upper branch of the bifurcation diagram until the cup disappears at $-h_{crit}$.

```
h_values = np.linspace(0, -0.5, 2000); x = 1.0; states = []
for h in h_values:
    x = F_tipmod(x, h, r=1, tau=2); states.append(x)
plot_analytical_bifurcation_tipmod(r=1, tau=2, extent=0.6); plt.plot(h_values, states, 'k', lw=2);
```

### Noise-induced tipping

Noise-induced tipping occurs throughout the middle part of this lecture: a single pulse larger than the width of the cup and a single random shock in the section on basin stability. In both, $h$ is fixed, and the cup exists.

### Rate-induced tipping

To illustrate the phenomenon of rate-induced tipping, we make the external conditions move the cup.

#### A moving cup

How can we formalize changing external drivers? We let them **move the cup**. The state favored by the current conditions, $\lambda_t$, moves by $\mu$ per time step, e.g., the sea level to which the ground level of a mangrove forest is adjusted. The system follows the moving cup with its response time $\tau$,

$$x_{t+1} = x_t + \frac{1}{\tau}\left(h + r(x_t-\lambda_t) - (x_t-\lambda_t)^3\right) + n\eta_t, \qquad \lambda_{t+1} = \lambda_t + \mu.$$

**Can the system keep up?**

We simulate the position of the system relative to the cup, $x_t - \lambda_t$.

```
def simulate_moving_cup(mu, h=0, r=1, tau=2, n=0, iters=150, xinit=1.0, seed=0):
    rng = np.random.default_rng(seed)  # random number generator; the seed makes the result reproducible
    x, lam = xinit, 0.0
    positions = [x - lam]  # position of the system relative to the cup
    for t in range(iters):
        x = F_tipmod_noise(x - lam, h, r, tau, n, rng) + lam  # the system follows the cup
        lam = lam + mu  # the cup moves
        positions.append(x - lam)
        if np.abs(x - lam) > 3: break  # stop the simulation when the system has left the cups
    return np.array(positions)
```

```
for mu in [0.1, 0.19, 0.2]:
    plt.plot(simulate_moving_cup(mu), label=f'$\\mu={mu}$')
plt.xlabel('Time steps $t$'); plt.ylabel('Position $x_t - \\lambda_t$'); plt.legend();
```

For slow change, the system stays in its cup but **lags behind**. For $\mu=0.2$, it falls behind so far that it drops into the other cup. Between $\mu=0.19$ and $\mu=0.2$ lies a sharp threshold.

What does this look like in the original coordinate $x$? The potential at time $t$ is $G(x - \lambda_t)$: the cups keep their shape and move to the right by $\mu$ per time step.

```
def plot_moving_cup(mu, times=(0, 15, 30, 40), h=0, r=1, tau=2):
    positions = simulate_moving_cup(mu, h=h, r=r, tau=tau, iters=max(times))
    x_grid = np.linspace(-2, 2, 401)  # coordinates relative to the cup
    for t in times:
        lam = mu*t  # position of the cup at time t
        line, = plt.plot(x_grid + lam, G_tipmod(x_grid, h, r, tau), label=f'$t={t}$')
        plt.plot(positions[t] + lam, G_tipmod(positions[t], h, r, tau), 'o', ms=10, color=line.get_color())
    plt.ylim(-0.3, 0.8); plt.xlabel('System state $x$'); plt.ylabel(r'Potential $G(x - \lambda_t)$'); plt.legend(ncol=4, loc='upper center');
```

```
plot_moving_cup(mu=0.2)
```

The system lags further and further behind the bottom of its cup, until it falls over the left edge into the other cup, which moves along.

#### Critical rate

Where exactly is the threshold? We describe the system relative to the cup, $y_t = x_t - \lambda_t$. Without shocks,

\begin{align}
y_{t+1} &= x_{t+1} - \lambda_{t+1} \\
&= x_t + \frac{1}{\tau}\left(h + ry_t - y_t^3\right) - \lambda_t - \mu \\
&= y_t + \frac{1}{\tau}\left((h - \tau\mu) + ry_t - y_t^3\right).
\end{align}

The moving cup acts like a **fixed cup with the external influence $h - \tau\mu$**. Seen from the moving cup, the shift becomes a tilt: $G(y;\, h - \tau\mu) = G(y;\, h) + \mu y$, a tilt with slope $\mu$.

For $\mu = 0.19$, the upper cup has almost disappeared:

```
plot_tipmod_potential(h=0 - 2*0.19)  # h - tau*mu for h=0, tau=2, mu=0.19
```

From [Tipping Elements](https://wbarfuss.github.io/csm-of-hei/02.02-TippingElements.html), the upper cup exists as long as the external influence stays above $-h_{crit}$, with $h_{crit} = \frac{2r}{3}\sqrt{\frac{r}{3}}$. Thus, the system keeps up as long as

$$h - \tau\mu > -h_{crit} \quad \Leftrightarrow \quad \mu < \frac{h + h_{crit}}{\tau} = \mu_{crit}.$$

For $h=0$, $r=1$, and $\tau=2$, $\mu_{crit} = 0.385/2 \approx 0.192$, the threshold we observed.

#### Faster response

The critical rate is inversely proportional to the response time $\tau$: **a system that responds faster can keep up with faster change**. Does this hold without limits? We find the critical rate by simulation, halving the interval between a rate that tips and one that does not.

```
def tips(mu, tau, h=0, r=1, iters=2000):
    positions = simulate_moving_cup(mu, h=h, r=r, tau=tau, iters=iters, xinit=1.01)  # start slightly off the equilibrium
    return len(positions) < iters+1 or np.any(positions < 0)  # left the upper cup at some time

def critical_rate(tau, h=0, r=1):
    lo, hi = 0.0, 2.0
    if tips(lo, tau, h, r): return 0.0  # the system leaves its cup even without change
    for _ in range(20):  # bisection
        mid = (lo + hi)/2
        if tips(mid, tau, h, r): hi = mid
        else: lo = mid
    return hi
```

```
taus = np.arange(0.5, 6.01, 0.1)
plt.plot(taus, [critical_rate(tau) for tau in taus], '.', label='Simulation')
plt.plot(taus, h_crit()/taus, 'k-', lw=1, label='$\\mu_{crit} = h_{crit}/\\tau$')
plt.axvline(1, color='k', ls=':', lw=1)  # tau = r
plt.xlabel('Response time $\\tau$'); plt.ylabel('Critical rate $\\mu_{crit}$'); plt.legend();
```

For $\tau \geq r$, the simulation follows $\mu_{crit} = h_{crit}/\tau$. A system that responds too fast **overreacts**: for $\tau < 2r$, it overshoots its equilibrium point but still returns to it; for $\tau < r$, the equilibrium point is unstable, and the state oscillates around it; for $\tau \lesssim 0.6$, it leaves its cup even when the conditions do not change (see the DeepDive on $\tau$ in [Tipping Elements](https://wbarfuss.github.io/csm-of-hei/02.02-TippingElements.html)).

#### Wider cup

The ball-and-cup diagram of adaptation shows a cup that becomes **wider**. In our model, $r$ sets the width: for $h=0$, the stable equilibrium point is $x_s = \sqrt{r}$ and the edge of the cup is $x_u = 0$. A wider cup raises the critical rate,

$$\mu_{crit} = \frac{h + h_{crit}}{\tau}, \qquad h_{crit} = \frac{2r}{3}\sqrt{\frac{r}{3}},$$

which grows like $r^{3/2}$. But a larger $r$ also moves the stability limit, $\tau > r$. **Does a wider cup pay off if it requires a slower response?**

We compare a narrow cup with a fast response, $r=1$ and $\tau=2$, with a cup of twice the strength whose response is twice as slow, $r=2$ and $\tau=4$. The cup moves with $\mu=0.23$.

```
for r, tau in [(1, 2), (2, 4)]:
    plt.plot(simulate_moving_cup(0.23, r=r, tau=tau, xinit=np.sqrt(r)), label=f'$r={r}$, $\\tau={tau}$')
plt.xlabel('Time steps $t$'); plt.ylabel('Position $x_t - \\lambda_t$'); plt.legend();
```

The wider cup keeps up, although it responds twice as slowly: its critical rate is $\mu_{crit} \approx 0.272$, compared with $0.192$ for the narrow cup. **A wider cup allows a proportionally slower response.** Doubling $r$ and $\tau$ together still raises $\mu_{crit}$ by the factor $2^{3/2}/2 = \sqrt{2}$ (see below).

#### Why the limits depend only on $r/\tau$

We measure the system state in units of the width of the cup, $x = \sqrt{r}\,u$. Inserting this into the update, step by step,

\begin{align}
\sqrt{r}\,\Delta u &= \frac{1}{\tau}\left(h + r\sqrt{r}\,u - r\sqrt{r}\,u^3\right) \\
\Delta u &= \frac{r}{\tau}\left(\frac{h}{r^{3/2}} + u - u^3\right).
\end{align}

The rescaled dynamics depend only on two numbers: the response speed relative to the reinforcing loop, $r/\tau$, and the external influence relative to the width, $h/r^{3/2}$. Thus, every limit for overreaction is a limit on $r/\tau$, for every $r$: the state overshoots for $\tau/r < 2$, the equilibrium point is unstable for $\tau/r < 1$, and the system leaves its cup without any change for $\tau/r \lesssim 0.6$. At the stability limit $\tau = r$, the critical rate is $\mu_{crit} = \frac{2}{3\sqrt{3}}\sqrt{r} \approx 0.385\sqrt{r}$: it still grows with the width $\sqrt{r}$ of the cup.

#### Moving cup and shocks

With shocks, the system has to keep up with the moving cup and absorb the shocks at the same time. We count how many of 200 simulations end in the lower cup, for $\mu=0.05$ and $n=0.1$.

```
def tipping_fraction(tau, mu=0.05, n=0.1, iters=1000, nseeds=200):
    tipped = 0
    for seed in range(nseeds):
        positions = simulate_moving_cup(mu, tau=tau, n=n, iters=iters, seed=seed)
        tipped += len(positions) < iters+1 or np.any(positions < 0)  # left the upper cup
    return tipped/nseeds

taus = [0.8, 1.0, 1.2, 1.5, 2, 3, 4, 6]
plt.plot(taus, [tipping_fraction(tau) for tau in taus], 'o-')
plt.xlabel('Response time $\\tau$'); plt.ylabel('Fraction of runs that tip');
```

A slow system lags behind the moving cup, which leaves a narrower margin for the shocks. A fast system with $\tau \leq r$ overreacts: its equilibrium point is unstable or marginally stable, the state oscillates within the cup, and the shocks push it over the edge. For all $\tau$ from 1.2 to 2, no run tips. **The system keeps up with the moving cup best at an intermediate response time.**
