## The subcritical pitchfork bifurcation

Bifurcation theory distinguishes different types of bifurcations. In the lecture, we met the fold bifurcation. A **subcritical pitchfork bifurcation** is one where an unstable equilibrium point splits into three, two unstable and one stable. Its difference equation is

$$\Delta x =x^3 - cx,$$

where $c$ is a parameter that controls the system's stability and $x$ is the system state[^pitchfork].

[^pitchfork]: The normal form of the subcritical pitchfork bifurcation is $\dot x = rx + x^3$ [@Strogatz1994, Sect. 3.4]. Our equation is this model in discrete time, with $r = -c$. We use $c$ because a larger $c$ makes the equilibrium point $x_e = 0$ more stable.
